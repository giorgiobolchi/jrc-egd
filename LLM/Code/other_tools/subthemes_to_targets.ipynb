This code is used to re-connect the subthemes to each targets after the target_to_target network has been regenerate dand aggregated

```python
# Import libraries
import sys
import pandas as pd
import numpy as np


# Define the project root directory
root_dir = '/Users/giorgiobolchi2/Documents/GitHub/jrc-egd/LLM/'

# Define absolute python path
sys.path.insert(0, root_dir) 
```

```python
# Data

network_data = pd.read_excel(f'{root_dir}Data/Outputs/0421/0421_network_aggregated.xlsx')
target_data = pd.read_csv(f'{root_dir}Data/targets_data_150.csv', sep=';')

output_dir = f'{root_dir}Data/Outputs/0421/'
```

```python


for i in range(len(network_data)): 
    match = target_data.loc[target_data['target_code'] == network_data.loc[i, 'source_target'], 'sub_theme'] 
    if not match.empty:
        network_data.loc[i, 'source_target_subtheme'] = match.iloc[0]

    if 'impact_target' in network_data.columns:
        match2 = target_data.loc[target_data['target_code'] == network_data.loc[i, 'impact_target'], 'sub_theme']
        if not match2.empty:
            network_data.loc[i, 'impact_target_subtheme'] = match2.iloc[0]

network_data.to_csv(path_or_buf= f'{output_dir}network_aggregated.csv', 
                    sep=';', 
                    index=False)
```
